# **归一化和标准化**

归一化和标准化的目的，就是将不同特征的数据缩放到一个统一的、可比的比例尺度上,以提升模型效果

## **归一化**

通过对原始数据进行变换,把数据映射到(min,max)(一般为[0,1]之间)

1. $$
X' = \frac{X-X_{min}}{X_{max}-X_{min}}
$$

2. $$
X''=X'*(max-min)+min
$$

$X$: 当前值

$X_{min}$: 该列最小值

$X_{max}$: 该列最大值

**注意**: 公式1极容易受到极值的影响,无论数据集多大

**适用场景**:
- 当特征的分布边界明确时：例如，图像的像素值天然就在 [0, 255] 之间，归一化到 [0, 1] 非常合适。
- 需要数据在固定范围内时：例如，某些神经网络期望的输入值是 0 到 1 之间。
- 数据分布不服从正态分布时。

> | 特征1 | 特征2 | 特征3 | 特征4 |
> | ----- | ----- | ----- | ----- |
> | 90    | 2     | 10    | 40    |
> | 60    | 4     | 15    | 45    |
> | 75    | 3     | 13    | 46    |
>
> 当对特征1进行归一化且min=0,max=1,套用公式：
>
> 特征1第一行
>
> $ X' = \frac{90-60}{90-60} = 1 $
>
> $ X'' = 1*(1-0)+0 = 1 $
>
> 特征1第二行
>
> $ X' = \frac{60-60}{90-60} = 0 $
>
> $ X'' = 0*(1-0)+0 = 0 $
>
> 特征1第三行
>
> $ X' = \frac{75-60}{90-60} = \frac{15}{30} = 0.5 $
>
> $ X'' = 0.5*(1-0)+0 = 0.5 $
>
> 假设min=3,max=5
>
>  $ X' = \frac{90-60}{90-60} = 1 $
>
>  $ X'' = 1*(5-3)+3 = 5 $



```python
from sklearn.preprocessing import MinMaxScaler # 归一化
# 1. 归一化之前的数据
x_train = [[90, 2, 10, 40], [60, 4, 15, 45], [75, 3, 13, 46]]
# 2. 创建归一化模型
transfer = MinMaxScaler()
# 3. 对原数据集进行归一化
x_train_minmax = transfer.fit_transform(x_train)
print( x_train_minmax)
```

## **标准化**

对原始数据进行标准化,转换后的数据服从均值为0,标准差为1的标准正态分布.

> **均值**: 所有数据的总和除以个数
>
> $ \mu = \frac{\sum_{i=1}^{n}x_i}{n}$
>
> **方差**: 每个数据与均值的差值的平方和,再除以数据个数(总体方差)或数据个数减 1(样本方差)
>
> $ \sigma^2 = \frac{\sum_{i=1}^{n}(x_i-\mu)^2}{n}$
>
> **标准差**: 方差的平方根
>
> $ \sigma = \sqrt{\frac{\sum_{i=1}^{n}(x_i-\mu)^2}{n}}$

$ X' = \frac{X-\mu}{\sigma}$

$ \mu$ : 特征的均值

$ \sigma$ : 特征的标准差

**适用场景**:
- 当特征的分布服从或近似正态分布时(但这不是强制要求)
- 数据中存在异常值时,标准化通常比归一化表现更好.
- 适用于大多数机器学习算法,特别是那些基于距离计算,假设数据来自高斯分布或使用梯度下降的算法.这几乎是机器学习中的"默认"预处理选择.

> | 特征1 | 特征2 | 特征3 | 特征4 |
> | ----- | ----- | ----- | ----- |
> | 90    | 2     | 10    | 40    |
> | 60    | 4     | 15    | 45    |
> | 75    | 3     | 13    | 46    |
>
> 当对特征1进行标准化时：
>
> 1. 计算该列的均值
>
> $ \mu = \frac{90+60+75}{3} = 75 $
>
> 2. 计算该列的方差
>
> $ \sigma^2 = \frac{(90-75)^2+(60-75)^2+(75-75)^2}{3} = 150 $
>
> 3. 计算该列的标准差
>
> $ \sigma = \sqrt{150} = 12.25 $
>
> 4. 对该列进行标准化
>
> 特征1第一行
>
> $ X' = \frac{90-75}{12.25} = \frac{15}{12.25} = 1.224 $
>
> 特征1第二行
>
> $ X' = \frac{60-75}{12.25} = \frac{-15}{12.25} = -1.224 $
>
> 特征1第三行
>
> $ X' = \frac{75-75}{12.25} = 0 $
>



```python
from sklearn.preprocessing import StandardScaler #标准化
# 1. 标准之前的数据
x_train = [[90, 2, 10, 40], [60, 4, 15, 45], [75, 3, 13, 46]]
# 2. 创建标准化模型
transfer = StandardScaler()
# 3. 对原数据集进行标准化
x_train_minmax = transfer.fit_transform(x_train)
print( x_train_minmax)
print('数据集的均值为:\n', transfer.mean_)
print('数据集的方差为:\n', transfer.var_)
print('数据集的标准差为:\n', transfer.scale_)
```
